# Module 3: Logic Gates with NumPy

This notebook demonstrates the `LogicGate` class defined in `logic_gate.py`. The class implements the **AND, NAND, OR, NOR and XOR** gates using NumPy's element-wise logical functions.

## 1. Imports

We import NumPy and the `LogicGate` class from our module.

```python
import numpy as np
from logic_gate import LogicGate

gate = LogicGate()
```

## 2. Implementation overview

Each gate is a static method that:

1. Validates that the inputs only contain `0`/`1` (or `True`/`False`).
2. Converts them to boolean NumPy arrays.
3. Applies a NumPy function (`np.logical_and`, `np.logical_or`, `np.logical_xor`, `np.logical_not`).
4. Returns the result as integers (`0` or `1`).

Because NumPy works element-wise, every gate accepts scalars and arrays alike.

## 3. Test inputs

All four possible combinations of two binary inputs.

```python
a = np.array([0, 0, 1, 1])
b = np.array([0, 1, 0, 1])
print("a =", a)
print("b =", b)
```

## 4. AND gate

Output is `1` only when **both** inputs are `1`.

```python
print("and_gate :", gate.and_gate(a, b))
print(LogicGate.truth_table(gate.and_gate))
```

## 5. NAND gate

The negation of AND: output is `0` only when both inputs are `1`.

```python
print("nand_gate:", gate.nand_gate(a, b))
print(LogicGate.truth_table(gate.nand_gate))
```

## 6. OR gate

Output is `1` when **at least one** input is `1`.

```python
print("or_gate  :", gate.or_gate(a, b))
print(LogicGate.truth_table(gate.or_gate))
```

## 7. NOR gate

The negation of OR: output is `1` only when **both** inputs are `0`.

```python
print("nor_gate :", gate.nor_gate(a, b))
print(LogicGate.truth_table(gate.nor_gate))
```

## 8. XOR gate

Output is `1` when **exactly one** input is `1`.

```python
print("xor_gate :", gate.xor_gate(a, b))
print(LogicGate.truth_table(gate.xor_gate))
```

## 9. Verifying against expected results

We assert that each gate matches its known truth table for inputs `(0,0), (0,1), (1,0), (1,1)`.

```python
expected = {
    "and_gate":  [0, 0, 0, 1],
    "nand_gate": [1, 1, 1, 0],
    "or_gate":   [0, 1, 1, 1],
    "nor_gate":  [1, 0, 0, 0],
    "xor_gate":  [0, 1, 1, 0],
}

for name, exp in expected.items():
    result = getattr(gate, name)(a, b)
    assert np.array_equal(result, exp), f"{name} failed"
    print(f"{name:10s} OK")
```

## 10. Input validation

Values other than `0`/`1` are rejected with a `ValueError`.

```python
try:
    gate.and_gate(2, 1)
except ValueError as err:
    print("Rejected invalid input:", err)
```

## 11. Summary

All five gates (`and_gate`, `nand_gate`, `or_gate`, `nor_gate`, `xor_gate`) are implemented with NumPy and produce the correct truth tables.
