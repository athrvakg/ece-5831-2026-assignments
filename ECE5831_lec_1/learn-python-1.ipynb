## 5. String Formatting

```python
name = "Atharva"
print("Hello, %s!" % name)

name = "Mary"
age = 30
print("%s is %d years old." % (name, age))

# f-strings (modern equivalent)
print(f"{name} is {age} years old.")

data = {"name": "Atharva", "age": "36"}
print("Name: %(name)s Age: %(age)s" % data)
```

## 7. Conditions

```python
x = 2
if x == 2:
    print("x equals 2")
elif x == 3:
    print("x equals 3")
else:
    print("x is neither 2 nor 3")

name = "Atharva"
age = 23
if name == "Atharva" and age == 23:
    print("Your name is Atharva, and you are also 23 years old.")

if name == "Atharva" or name == "Rick":
    print("Your name is either Atharva or Rick.")
```

## 8. Loops

```python
count = 0
while count < 5:
    print(count)
    count += 1

for x in range(5):
    print(x)

for x in range(3, 8):
    print(x)

for x in range(2, 10, 2):
    print(x)

for x in range(5):
    if x == 2:
        break
    print("x is currently:", x)

for x in range(10):
    if x % 2 == 0:
        continue
    print(x)
```

## 9. Functions

```python
def my_function():
    print("Hello From My Function!")

def my_function_with_args(username, greeting):
    print(f"Hello, {username}, From My Function! I wish you {greeting}")

def sum_two_numbers(a, b):
    return a + b

my_function()
my_function_with_args("Atharva", "a great year!")
x = sum_two_numbers(1, 2)
print(x)
```

## 10. Classes and Objects

```python
class MyClass:
    variable = "blah"

    def function(self):
        print("This is a message inside the class.")

myobjectx = MyClass()
print(myobjectx.variable)
myobjectx.function()

class NumberHolder:
    def __init__(self, number):
        self.number = number

    def add_number(self, number):
        self.number += number

my_number_holder = NumberHolder(23)
my_number_holder.add_number(20)
print(my_number_holder.number)
```

## 11. Dictionaries

```python
phonebook = {}
phonebook["Atharva"] = 938477566
phonebook["Jack"] = 938377264
phonebook["Jill"] = 947662781
print(phonebook)

phonebook = {
    "Atharva": 938477566,
    "Jack": 938377264,
    "Jill": 947662781
}
for name, number in phonebook.items():
    print(f"Phone number of {name} is {number}")

del phonebook["Atharva"]
print(phonebook)
phonebook.pop("Jack")
print(phonebook)
```
