# Section 12. Working with Files

### Setup: a scratch working directory

```python
import os, tempfile
work_dir = tempfile.mkdtemp(prefix="pytut_")
os.chdir(work_dir)
print("Working in:", work_dir)
```

### Create a New Text File

```python
with open("notes.txt", "w") as f:
    f.write("")  # creates an empty file
print("notes.txt" in os.listdir("."))
```

### Write to a Text File

```python
with open("notes.txt", "w") as f:
    f.write("Hello, this is line 1.\n")
    f.write("This is line 2.\n")
print("Wrote to notes.txt")
```

### Read from a Text File

```python
with open("notes.txt", "r") as f:
    content = f.read()
print(content)

with open("notes.txt", "r") as f:
    for line in f:
        print("Line:", line.strip())
```

### Check if a File Exists

```python
print(os.path.exists("notes.txt"))
print(os.path.exists("does_not_exist.txt"))
```

### Write CSV Files

```python
import csv
with open("people.csv", "w", newline="") as f:
    writer = csv.writer(f)
    writer.writerow(["name", "age"])
    writer.writerow(["Alice", 30])
    writer.writerow(["Bob", 25])
print("Wrote people.csv")
```

### Read CSV Files

```python
with open("people.csv", "r", newline="") as f:
    reader = csv.reader(f)
    for row in reader:
        print(row)
```

### Rename a File

```python
os.rename("notes.txt", "notes_renamed.txt")
print(sorted(os.listdir(".")))
```

### Delete a File

```python
os.remove("notes_renamed.txt")
print(sorted(os.listdir(".")))
```

# Section 13. Working Directories

### Working with Directories

```python
print("Current directory:", os.getcwd())

os.mkdir("subdir")
print("subdir" in os.listdir("."))

os.rmdir("subdir")
print("subdir" in os.listdir("."))
```

### List Files in a Directory

```python
with open("a.txt", "w") as f:
    f.write("a")
with open("b.txt", "w") as f:
    f.write("b")

print("All entries:", sorted(os.listdir(".")))

for entry in sorted(os.listdir(".")):
    kind = "dir" if os.path.isdir(entry) else "file"
    print(entry, "->", kind)

os.remove("a.txt")
os.remove("b.txt")
os.remove("people.csv")
```

# Section 14. Strings

### F-strings

```python
name = "Alice"
age = 30
print(f"{name} is {age} years old.")
print(f"Next year, {name} will be {age + 1}.")

pi = 3.14159265
print(f"Pi rounded: {pi:.2f}")
```

### Raw Strings

```python
path = r"C:\Users\name\Documents"
print(path)

normal = "C:\\Users\\name"  # escaped backslashes without a raw string
print(normal)
```

### Backslash

```python
print("Tab:\tEnd")
print("Newline:\nEnd")
print("Backslash: \\")
print('Quote inside string: \'single\' and \"double\"')
```

# Section 15. Third-party Packages, PIP, and Virtual Environments

### Python Package Index (PyPI) and pip

```python
# pip is Python's package installer, used from the command line/terminal:
#   pip install <package-name>
#   pip install pandapower[all]
#   pip list
#   pip show <package-name>
#   pip uninstall <package-name>
#
# PyPI (pypi.org) is the online repository pip downloads packages from.
print("pip is run from the terminal, e.g.: pip install requests")

# We can check installed package info programmatically too:
import importlib.metadata as im
try:
    print("numpy version:", im.version("numpy"))
except im.PackageNotFoundError:
    print("numpy not installed in this environment")
```

### Virtual Environments

```python
# A virtual environment is an isolated Python environment with its own
# installed packages, separate from the system/global Python installation.
# This avoids version conflicts between different projects.
#
# Creating one with venv (standard library):
#   python -m venv myenv
#
# Activating it:
#   Windows:      myenv\Scripts\activate
#   macOS/Linux:  source myenv/bin/activate
#
# With conda (what we used for this course's ece-5831-2026 environment):
#   conda create -n ece-5831-2026 python=3.11
#   conda activate ece-5831-2026
print("Virtual environments isolate per-project dependencies.")
print("This course uses a conda environment: ece-5831-2026")
```

### Install pipenv on Windows

```python
# pipenv combines pip + virtualenv into one tool. To install it on Windows:
#   pip install pipenv
#
# Then create/activate a pipenv-managed environment for a project:
#   pipenv install        # creates a Pipfile-based virtual environment
#   pipenv shell          # activates it
print("pipenv is installed via: pip install pipenv")
```

### Cleanup

```python
import shutil
os.chdir(tempfile.gettempdir())
shutil.rmtree(work_dir, ignore_errors=True)
print("Cleaned up scratch directory.")
```
