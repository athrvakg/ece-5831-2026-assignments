## 1. Regular Expressions

```python
import re

pattern = r"^[a-zA-Z0-9-_]*$"
print(bool(re.match(pattern, "username-123")))
print(bool(re.match(pattern, "invalid username!")))

text = "cat hat bat mat"
matches = re.findall(r"[a-z]at", text)
print(matches)

result = re.sub(r"at", "og", text)
print(result)
```

## 2. Exception Handling

```python
def do_stuff_with_number(n):
    print(n)

def catch_this():
    the_list = (1, 2, 3, 4, 5)
    for i in range(6):
        try:
            do_stuff_with_number(the_list[i])
        except IndexError:
            do_stuff_with_number(0)

catch_this()

try:
    x = 1 / 0
except ZeroDivisionError as e:
    print("Caught a division error:", e)
finally:
    print("This runs no matter what.")
```

## 3. Sets

```python
a = set(["Jake", "John", "Eric"])
b = set(["John", "Jill"])

print(a)
print(a & b)          # intersection
print(a | b)          # union
print(a - b)          # difference
print(a ^ b)          # symmetric difference

a.add("Ann")
print(a)
print("Jake" in a)
a.remove("Ann")
print(a)
```

## 4. Serialization

```python
import json

my_list = ["foo", "bar", "baz"]
json_string = json.dumps(my_list)
print(json_string)

my_dict = json.loads(json_string) if isinstance(json.loads(json_string), dict) else json.loads(json_string)
print(my_dict)

data = {"a": 1, "b": 2, "c": [1, 2, 3]}
s = json.dumps(data)
print(s)
print(json.loads(s))
```

## 5. Partial Functions

```python
from functools import partial

def multiply(x, y):
    return x * y

double = partial(multiply, 2)
triple = partial(multiply, 3)

print(double(8))
print(triple(8))
```

## 6. Code Introspection

```python
def my_func(a, b=2):
    """An example function for introspection."""
    return a + b

print(my_func.__doc__)
print(my_func.__name__)
help(my_func)
```

## 7. Closures

```python
def transmit_to_space(message):
    "This is the enclosing function"
    def data_transmitter():
        "The nested function"
        print(message)
    return data_transmitter

fun2 = transmit_to_space("Burn the Sun!")
fun2()
```

## 8. Decorators

```python
def repeater(old_function):
    def new_function(*args, **kwds):
        old_function(*args, **kwds)
        old_function(*args, **kwds)
    return new_function

@repeater
def multiply(num1, num2):
    print(num1 * num2)

multiply(2, 3)
```

## 9. Generators

```python
def double_numbers(iterable):
    for i in iterable:
        yield i + i

for i in double_numbers(range(1, 5)):
    print(i)

gen = (x * x for x in range(5))
print(list(gen))
```

## 10. List Comprehensions

```python
sentence = "the quick brown fox jumps over the lazy dog"
words = sentence.split()
word_lengths = [len(word) for word in words if word != "the"]
print(words)
print(word_lengths)

squares = [x**2 for x in range(10)]
print(squares)
```

## 11. Multiple Function Arguments

```python
def foo(first, second, third, *therest):
    print(f"First: {first}")
    print(f"Second: {second}")
    print(f"Third: {third}")
    print(f"And all the rest... {therest}")

foo(1, 2, 3, 4, 5)

def bar(first, second, third, **options):
    if options.get("action") == "sum":
        print(f"The sum is: {first + second + third}")
    if options.get("number") == "first":
        return first

result = bar(1, 2, 3, action="sum", number="first")
print(result)
```

## 12. Multiple Inheritance

```python
class Animal:
    def __init__(self, name):
        self.name = name

    def speak(self):
        raise NotImplementedError("Subclass must implement abstract method")

class Fish(Animal):
    def speak(self):
        return "Blub"

class Guitar:
    def play(self):
        return "Strum!"

class WeirdCreature(Fish, Guitar):
    pass

wc = WeirdCreature("Chimera")
print(wc.speak())
print(wc.play())
```
